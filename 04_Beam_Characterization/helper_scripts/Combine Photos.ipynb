```python
from PIL import Image
import math
from pathlib import Path

# ── Configuration ──────────────────────────────────────────────────────────────
INPUT_FOLDER = r"C:\Users\ikcco\OneDrive\Desktop\IsaacCross_Thesis_Stuff\Thesis Figures\ch06_results\Summary Plots Good" # folder containing your PNGs
OUTPUT_FILE  = r"C:\Users\ikcco\OneDrive\Desktop\IsaacCross_Thesis_Stuff\Thesis Figures\ch06_results\summary_plots_good_combined.png"
COLS         = 2          # images per row
# ──────────────────────────────────────────────────────────────────────────────

paths = sorted(Path(INPUT_FOLDER).glob("*.png"))
if not paths:
    raise FileNotFoundError(f"No PNG files found in {INPUT_FOLDER!r}")

imgs = [Image.open(p) for p in paths]
w, h = imgs[0].size          # assumes all images are the same size

rows = math.ceil(len(imgs) / COLS)
canvas = Image.new("RGB", (w * COLS, h * rows), color=(255, 255, 255))

for i, img in enumerate(imgs):
    col = i % COLS
    row = i // COLS
    canvas.paste(img, (col * w, row * h))

canvas.save(OUTPUT_FILE)
print(f"Saved {len(imgs)} images → {rows}×{COLS} grid → {OUTPUT_FILE}")
```
